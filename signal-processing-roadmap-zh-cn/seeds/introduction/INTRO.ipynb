# 信号处理路线图

这套课程从一个简单问题开始：**一段随时间变化的数，怎样变成可以观察、解释和使用的信息？**

不需要先学完复杂数学，我们从能够看见、能够画出的信号开始。


## 信号是什么？

信号是随时间或空间变化的量。声音的空气压强、心电图的电压、雷达回波、传感器读数，都可以记录成信号。

信号处理通常沿着这条路径工作：

1. 生成或采集
2. 观察
3. 变换
4. 分析
5. 判断或应用

## 课程地图

16 周大致分为两段：

- **基础**：信号与时间轴、采样、傅里叶、滤波、频谱
- **应用**：音频、生物医学、雷达与通信、物联网、金融等

第 1 周依次学习信号与时间轴、生成简单信号、信号组合与变换，最后做一次综合练习。

### 怎么学

- 每个子章节是一个学习单元，有自己的 Notebook、Kernel 和工作区。
- 跟着 Step 看、改、运行；需要动手的练习要自己完成后再继续。
- Stop 后可以从当前步骤重放，不需要从头开始。
- 本课程不使用自动判分；图表和结果由你自己观察。
