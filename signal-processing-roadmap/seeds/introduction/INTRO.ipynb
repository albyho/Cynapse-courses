# Signal Processing Roadmap

This course opens with one simple question: **how does a series of values that changes over time become information we can observe, explain and use?**

We do not need the full mathematics first. We start with signals we can see and draw.


## What is a signal?

A signal is a quantity that changes over time or space. Air pressure in sound, voltage in an ECG, radar echoes and sensor readings can all be recorded as signals.

Signal processing usually follows this path:

1. Generate or acquire
2. Observe
3. Transform
4. Analyze
5. Judge or apply


## Reference project

This course follows the chapter skeleton and topics of the open source project
[Signal processing roadmap](https://github.com/minthanthtoo/signal-processing-roadmap).

The original repository organizes a signal processing learning path into chapters and notebooks. This course keeps its topic order, but the walkthroughs, notebooks, exercises and figures here are all newly written; they are not a translation or a mirror of the source repository.


## Course map

The 16 weeks fall into two halves:

- **Foundations**: signals and the time axis, sampling, Fourier, filters, spectra
- **Applications**: audio, biomedical, radar and communications, IoT, finance, and more

Week 1 works through signals and the time axis, generating simple signals, combining and transforming them, and closes with one integrative exercise.

### How to learn

- Each subsection is a learning unit with its own notebook, kernel and workspace.
- Follow each step: read, edit, run. Hands-on exercises are yours to finish before moving on.
- After a Stop you can replay from the current step; there is no need to start over.
- This course does not grade automatically; you observe the figures and results yourself.
