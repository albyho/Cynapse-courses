# 1.2 Generating and Plotting Simple Signals

This section turns signal into a sine function we can generate, and watches three parameters.


## What this unit covers

- Generate a sine signal on a sample time axis
- Understand the amplitude `A`, the frequency `f`, the period `T`, the angular frequency `omega` and the initial phase `phi`
- Predict the waveform change from a parameter change, then verify it


## One formula to keep

$$x(t) = A \sin(2 \pi f t + \phi)$$

Do not derive it yet. Watch what the figure does as each parameter changes.
