# 1.4 Putting It All Together

Express two components in one signal, and explain your parameters in words.


## Task

Build a signal made of two sines:

- a low frequency component, around 1 Hz
- a high frequency component, around 4 Hz
- at least one component with a phase or a time shift

Draw the waveform and add a title, axis labels and a legend.


## There is no single answer

This course does not grade automatically. What matters is that you can say clearly:

1. which parameters you used
2. what you predicted
3. what you actually saw
