# 1.1 Signals and the Time Axis

Putting a signal on a time axis is the first step to observing it.


## What this unit covers

- Tell a continuous signal from a discrete one
- Get to know the sample index, the sample time and the amplitude
- Draw the same samples with `plot` and with `stem`, and explain how the two figures differ


## Observation task

The same values can be drawn as a polyline or as vertical lines. Which drawing fits the meaning of discrete samples?
