# 1.3 Combining and Transforming Signals

A signal can be observed, but also added, scaled, shifted and reversed.


## What this unit covers

- Add two signals sample by sample
- Scale the amplitude with a constant
- Shift time with `t - t0`
- Understand reversing the sample order


## Where these operations are used

Filtering, modulation and system analysis later all reuse these basic operations.

For now, watch the waveform only; no new formulas.
